```python
import pandas as pd

def answers_equal(a, b):
    if pd.isna(a) and pd.isna(b):
        return True
    if pd.isna(a) or pd.isna(b):
        return False
    return str(a).strip().casefold() == str(b).strip().casefold()

df['answer_match'] = df.apply(lambda r: answers_equal(r['extracted_answer'], r['answer']), axis=1)

total = len(df)
matches = int(df['answer_match'].sum())
print(f"Matches: {matches}/{total} ({matches/total:.0%})")

mismatches = df[~df['answer_match']]
if len(mismatches):
    display_cols = ['question', 'answer', 'extracted_answer', 'answerable', 'answer_start']
    print(f"\nMismatches ({len(mismatches)} rows):")
    print(mismatches[display_cols])
else:
    print("All rows match.")
```
